```python
import pandas as pd

file_path = "Dataset/sample_superstore _dataset.xls"

orders = pd.read_excel(file_path, sheet_name="Orders")
people = pd.read_excel(file_path, sheet_name="People")
returns = pd.read_excel(file_path, sheet_name="Returns")

print("Orders:", orders.shape)
print("People:", people.shape)
print("Returns:", returns.shape)
```

```python
# Create a copy of the Orders dataset
orders_clean = orders.copy()

# Remove accidental spaces from column names
orders_clean.columns = orders_clean.columns.str.strip()

# Check the cleaned dataset
print("Cleaned dataset shape:", orders_clean.shape)
print("\nMissing values:")
print(orders_clean.isnull().sum().sum())

print("\nDuplicate rows:")
print(orders_clean.duplicated().sum())
```

```python
numeric_columns = ["Sales", "Quantity", "Discount", "Profit"]

orders_clean[numeric_columns].describe()
```

```python
print("Categories:")
print(orders_clean["Category"].value_counts())

print("\nSub-Categories:")
print(orders_clean["Sub-Category"].nunique(), "unique sub-categories")

print("\nRegions:")
print(orders_clean["Region"].value_counts())

print("\nSegments:")
print(orders_clean["Segment"].value_counts())
```

```python
orders_clean["Year"] = orders_clean["Order Date"].dt.year
orders_clean["Month"] = orders_clean["Order Date"].dt.month
orders_clean["Month_Name"] = orders_clean["Order Date"].dt.month_name()

print(orders_clean[["Order Date", "Year", "Month", "Month_Name"]].head())
```

```python
total_sales = orders_clean["Sales"].sum()
total_profit = orders_clean["Profit"].sum()
total_quantity = orders_clean["Quantity"].sum()
total_orders = orders_clean["Order ID"].nunique()
total_customers = orders_clean["Customer ID"].nunique()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
print("Total Quantity:", total_quantity)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
```

```python
category_analysis = orders_clean.groupby("Category").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum"),
    total_orders=("Order ID", "nunique")
).reset_index()

category_analysis["profit_margin_percentage"] = (
    category_analysis["total_profit"] /
    category_analysis["total_sales"] * 100
).round(2)

category_analysis = category_analysis.sort_values(
    "total_profit", ascending=False
)

category_analysis
```

```python
subcategory_analysis = orders_clean.groupby(
    ["Category", "Sub-Category"]
).agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum"),
    total_orders=("Order ID", "nunique")
).reset_index()

subcategory_analysis["profit_margin_percentage"] = (
    subcategory_analysis["total_profit"] /
    subcategory_analysis["total_sales"] * 100
).round(2)

subcategory_analysis = subcategory_analysis.sort_values(
    "total_profit",
    ascending=False
)

subcategory_analysis
```

```python
region_analysis = orders_clean.groupby("Region").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum"),
    total_orders=("Order ID", "nunique"),
    total_customers=("Customer ID", "nunique")
).reset_index()

region_analysis["profit_margin_percentage"] = (
    region_analysis["total_profit"] /
    region_analysis["total_sales"] * 100
).round(2)

region_analysis = region_analysis.sort_values(
    "total_profit",
    ascending=False
)

region_analysis
```

```python
segment_analysis = orders_clean.groupby("Segment").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum"),
    total_orders=("Order ID", "nunique"),
    total_customers=("Customer ID", "nunique")
).reset_index()

segment_analysis["profit_margin_percentage"] = (
    segment_analysis["total_profit"] /
    segment_analysis["total_sales"] * 100
).round(2)

segment_analysis = segment_analysis.sort_values(
    "total_profit",
    ascending=False
)

segment_analysis
```

```python
# Return Analysis

print("Returns dataset shape:", returns.shape)
print("\nReturns columns:")
print(returns.columns.tolist())

print("\nFirst 5 returned orders:")
print(returns.head())
```

```python
# Calculate return rate

total_returned_orders = returns["Order ID"].nunique()
total_orders = orders_clean["Order ID"].nunique()

return_rate = (total_returned_orders / total_orders) * 100

print("Total Orders:", total_orders)
print("Total Returned Orders:", total_returned_orders)
print("Return Rate:", round(return_rate, 2), "%")
```

```python
# Returned orders by region

returned_orders = returns[returns["Returned"] == "Yes"]

returned_region_analysis = returned_orders.merge(
    orders_clean[["Order ID", "Region"]],
    on="Order ID",
    how="left"
)

returned_region_analysis = (
    returned_region_analysis.groupby("Region")
    .agg(returned_orders=("Order ID", "nunique"))
    .reset_index()
)

returned_region_analysis["return_rate_percentage"] = (
    returned_region_analysis["returned_orders"] / total_orders * 100
).round(2)

returned_region_analysis = returned_region_analysis.sort_values(
    "returned_orders",
    ascending=False
)

returned_region_analysis
```

```python
# Returned orders by category

returned_category_analysis = returned_orders.merge(
    orders_clean[["Order ID", "Category"]],
    on="Order ID",
    how="left"
)

returned_category_analysis = (
    returned_category_analysis.groupby("Category")
    .agg(returned_orders=("Order ID", "nunique"))
    .reset_index()
)

returned_category_analysis["return_rate_percentage"] = (
    returned_category_analysis["returned_orders"] / total_orders * 100
).round(2)

returned_category_analysis = returned_category_analysis.sort_values(
    "returned_orders",
    ascending=False
)

returned_category_analysis
```
